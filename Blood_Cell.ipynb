import necessary libraries

```python
import  cv2
import numpy as np
import ultralytics
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Start Recignition and Image Processing


```python
pic=cv2.imread('3.png')
plt.imshow(cv2.cvtColor(pic, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

Start With Otsu based Reconition Process

```python


gray=cv2.cvtColor(pic,cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(20,8))

plt.subplot(1, 3, 2)
plt.imshow(gray, cmap='gray')

plt.show()
```

Get Automated Treshold in Otsu Algorithm

```python

_, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

plt.figure(figsize=(13,8))
plt.subplot(1, 2, 2)
plt.imshow(thresh, cmap='gray')

plt.show()
```

All Contour Selection 

```python
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

result = pic.copy()
cv2.drawContours(result, contours, -1, (0, 255, 0), 2)

plt.figure(figsize=(7, 4))
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title(len(contours))
plt.axis('off')
plt.show()

print(len(contours))
```

Main Contour Selection

```python
filtered_contours = []
areas = []

for contour in contours:
    area = cv2.contourArea(contour)
    areas.append(area)

    if area >800: 
        filtered_contours.append(contour)

print(len(contours))
print(len(filtered_contours))


result_filtered = pic.copy()
cv2.drawContours(result_filtered, filtered_contours, -1, (0, 255, 0), 2)

plt.figure(figsize=(6, 8))
plt.imshow(cv2.cvtColor(result_filtered, cv2.COLOR_BGR2RGB))
plt.title(len(filtered_contours))
plt.axis('off')
plt.show()


print(len(contours))
print(len(filtered_contours))
```
